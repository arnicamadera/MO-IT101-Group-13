```python
# ============================================
# STEP 1: ANALYZE THE DATASETS
# ============================================

import pandas as pd
import numpy as np

# Load Product List dataset
product_df = pd.read_csv("MotorPH_Products_List_2025.csv")

# Load Sales dataset
sales_df = pd.read_csv("MotorPH_Sales Data-3rd Quarter-Year 2025.csv")

print("Datasets loaded successfully!")

# ============================================
# VIEW FIRST 5 RECORDS
# ============================================

print("===== PRODUCT LIST =====")
display(product_df.head())

print("\n===== SALES DATASET =====")
display(sales_df.head())

# ============================================
# CHECK DATASET STRUCTURE
# ============================================

print("===== PRODUCT LIST STRUCTURE =====")

print("Number of rows:", product_df.shape[0])
print("Number of columns:", product_df.shape[1])

print("\nColumn names:")
print(product_df.columns.tolist())

print("\nData types:")
print(product_df.dtypes)


print("\n\n===== SALES DATASET STRUCTURE =====")

print("Number of rows:", sales_df.shape[0])
print("Number of columns:", sales_df.shape[1])

print("\nColumn names:")
print(sales_df.columns.tolist())

print("\nData types:")
print(sales_df.dtypes)

# ============================================
# DETAILED DATASET INFORMATION
# ============================================

print("===== PRODUCT LIST INFORMATION =====")
product_df.info()

print("\n\n===== SALES DATASET INFORMATION =====")
sales_df.info()

# ============================================
# STATISTICAL SUMMARY
# ============================================

print("===== PRODUCT LIST STATISTICS =====")
display(product_df.describe(include="all"))

print("\n===== SALES DATASET STATISTICS =====")
display(sales_df.describe(include="all"))

# ============================================
# CHECK FOR MISSING VALUES
# ============================================

print("===== PRODUCT LIST MISSING VALUES =====")

product_missing = product_df.isnull().sum()

print(product_missing)
print("\nTotal missing values:",
      product_missing.sum())


print("\n\n===== SALES DATASET MISSING VALUES =====")

sales_missing = sales_df.isnull().sum()

print(sales_missing)
print("\nTotal missing values:",
      sales_missing.sum())

# ============================================
# DISPLAY ROWS WITH MISSING VALUES
# ============================================

print("===== PRODUCT LIST ROWS WITH MISSING VALUES =====")

display(
    product_df[
        product_df.isnull().any(axis=1)
    ]
)


print("\n===== SALES ROWS WITH MISSING VALUES =====")

display(
    sales_df[
        sales_df.isnull().any(axis=1)
    ]
)

# ============================================
# CHECK DUPLICATE RECORDS
# ============================================

print("===== PRODUCT LIST DUPLICATES =====")

print(
    "Number of duplicate rows:",
    product_df.duplicated().sum()
)


print("\n===== SALES DATASET DUPLICATES =====")

print(
    "Number of duplicate rows:",
    sales_df.duplicated().sum()
)

# ============================================
# CHECK DUPLICATE PRODUCT IDs
# ============================================

print("===== PRODUCT ID CHECK =====")

print(
    "Number of unique Product IDs:",
    product_df["EntrNo"].nunique()
)

print(
    "Total Product IDs:",
    len(product_df)
)

print("\nDuplicate Product IDs:")

display(
    product_df[
        product_df["EntrNo"].duplicated(keep=False)
    ]
)

# ============================================
# CHECK DATA TYPES
# ============================================

print("===== PRODUCT LIST DATA TYPES =====")

for column in product_df.columns:
    print(
        f"{column}: {product_df[column].dtype}"
    )


print("\n===== SALES DATA TYPES =====")

for column in sales_df.columns:
    print(
        f"{column}: {sales_df[column].dtype}"
    )

# ============================================
# CHECK DATE FORMATTING
# ============================================

print("===== SALES DATE CHECK =====")

print("Data type:")
print(sales_df["date"].dtype)

print("\nSample dates:")
print(sales_df["date"].head(10))

print("\nNumber of unique dates:")
print(sales_df["date"].nunique())

# ============================================
# CHECK CLIENT TYPE CATEGORIES
# ============================================

print("===== CLIENT TYPE CATEGORIES =====")

print(
    sales_df["client_type"].value_counts(
        dropna=False
    )
)

# ============================================
# CHECK PAYMENT CATEGORIES
# ============================================

print("===== PAYMENT CATEGORIES =====")

print(
    sales_df["payment"].value_counts(
        dropna=False
    )
)

# ============================================
# CHECK SALES PRODUCT NAMES
# ============================================

print("===== SALES PRODUCT NAMES =====")

print(
    "Number of unique product names:",
    sales_df["product"].nunique()
)

print("\nProduct names:")

display(
    pd.DataFrame(
        sorted(
            sales_df["product"]
            .dropna()
            .unique()
        ),
        columns=["Product Name"]
    )
)

# ============================================
# CHECK PRODUCT LIST NAMES
# ============================================

print("===== PRODUCT LIST PRODUCT NAMES =====")

print(
    "Number of unique products:",
    product_df["EntrName"].nunique()
)

display(
    product_df[
        ["EntrNo", "EntrName"]
    ]
)

# ============================================
# CHECK NUMERICAL VALUES
# ============================================

print("===== PRODUCT LIST NUMERICAL COLUMNS =====")

display(
    product_df[
        [
            "Manufacturing Date",
            "Acquisiton",
            "UnitPrice"
        ]
    ].describe()
)


print("\n===== SALES NUMERICAL COLUMNS =====")

display(
    sales_df[
        [
            "Unit Price",
            "quantity",
            "total"
        ]
    ].describe()
)

# ============================================
# CHECK SALES TOTAL CALCULATIONS
# ============================================

sales_df["Calculated Total"] = (
    sales_df["Unit Price"]
    * sales_df["quantity"]
)

incorrect_totals = sales_df[
    sales_df["Calculated Total"]
    != sales_df["total"]
]

print(
    "Number of records with incorrect totals:",
    len(incorrect_totals)
)

display(
    incorrect_totals[
        [
            "date",
            "product",
            "unit price",
            "quantity",
            "total",
            "Calculated Total"
        ]
    ].head(20)
)

# ============================================
# STEP 1 ANALYSIS SUMMARY
# ============================================

print("============================================")
print("       STEP 1: DATASET ANALYSIS SUMMARY")
print("============================================")

print("\nPRODUCT LIST")
print("--------------------------------------------")
print("Rows:", product_df.shape[0])
print("Columns:", product_df.shape[1])
print(
    "Missing values:",
    product_df.isnull().sum().sum()
)
print(
    "Duplicate rows:",
    product_df.duplicated().sum()
)
print(
    "Unique Product IDs:",
    product_df["EntrNo"].nunique()
)
print(
    "Unique Product Names:",
    product_df["EntrName"].nunique()
)


print("\nSALES DATASET")
print("--------------------------------------------")
print("Rows:", sales_df.shape[0])
print("Columns:", sales_df.shape[1])
print(
    "Missing values:",
    sales_df.isnull().sum().sum()
)
print(
    "Duplicate rows:",
    sales_df.duplicated().sum()
)
print(
    "Unique Products:",
    sales_df["product"].nunique()
)
unmatched_products = (
    set(sales_df["product"].dropna())
    - set(product_df["EntrName"].dropna())
)

print(
    "Unmatched Product Names:",
    len(unmatched_products)
)
print(
    "Incorrect Sales Totals:",
    len(incorrect_totals)
)


print("\n============================================")
print("Potential issues identified:")
print("============================================")
print("1. Missing values exist in the Sales dataset.")
print("2. Some columns require data type checking/conversion.")
print("3. Product names contain inconsistent/incorrect entries.")
print("4. Some sales totals do not match Unit Price × Quantity.")
print("5. Duplicate records were checked.")
print("6. Category values were reviewed for consistency.")

# ============================================
# STEP 2: DATA PREPROCESSING
# ============================================

# Create copies so the original datasets are preserved

products = product_df.copy()
sales = sales_df.copy()

print("Copies of the datasets created.")

# ============================================
# RENAME PRODUCT LIST COLUMNS
# ============================================

products = products.rename(columns={
    "EntrNo": "Product ID Number",
    "EntrName": "Product Name",
    "EntrDetails": "Product Details",
    "Manufacturing Date": "Manufacturing Year",
    "Acquisiton": "Acquisiton Year",
    "UnitPrice": "Unit Price"
})

print("Product List columns after renaming:")
print(products.columns.tolist())

# ============================================
# CLEAN PRODUCT LIST TEXT
# ============================================

products["Product Name"] = (
    products["Product Name"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

products["Product Details"] = (
    products["Product Details"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Product text values standardized.")

# ============================================
# CREATE PRODUCT TYPE
# ============================================

products["Product Type"] = (
    products["Product Details"]
    .str.split("/", n=1)
    .str[0]
    .str.strip()
    .str.title()
)

display(
    products[
        [
            "Product Name",
            "Product Type"
        ]
    ].head(10)
)

# ============================================
# CORRECT PRODUCT LIST DATA TYPES
# ============================================

products["Product ID Number"] = pd.to_numeric(
    products["Product ID Number"],
    errors="coerce"
)

products["Manufacturing Year"] = pd.to_numeric(
    products["Manufacturing Year"],
    errors="coerce"
).astype("Int64")

products["Acquisiton Year"] = pd.to_numeric(
    products["Acquisiton Year"],
    errors="coerce"
).astype("Int64")

products["Unit Price"] = pd.to_numeric(
    products["Unit Price"],
    errors="coerce"
)

print("Product List data types after conversion:")
print(products.dtypes)

# ============================================
# REMOVE DUPLICATE PRODUCT RECORDS
# ============================================

before = len(products)

products = products.drop_duplicates()

after = len(products)

print("Records before removing duplicates:", before)
print("Records after removing duplicates:", after)
print("Duplicate records removed:", before - after)

# ============================================
# MAKE PRODUCT IDs SEQUENTIAL
# ============================================

products = products.sort_values(
    "Product ID Number"
).reset_index(drop=True)

products["Product ID Number"] = range(
    1,
    len(products) + 1
)

print("Product IDs have been made sequential.")

display(
    products[
        [
            "Product ID Number",
            "Product Name"
        ]
    ].head(10)
)

# ============================================
# RENAME SALES COLUMNS
# ============================================

sales.columns = [
    "Date",
    "Client Type",
    "Product",
    "Unit Price",
    "Quantity",
    "Total",
    "Payment"
]

print("Sales columns after renaming:")
print(sales.columns.tolist())

# ============================================
# CLEAN SALES TEXT
# ============================================

for column in [
    "Client Type",
    "Product",
    "Payment"
]:

    sales[column] = (
        sales[column]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

print("Sales text values standardized.")

# ============================================
# CORRECT PRODUCT NAME INCONSISTENCIES
# ============================================

product_name_corrections = {
    "Bajaj CT12x": "Bajaj CT125",
    "Benelli 502x": "Benelli 502C",
    "Bristol Bobber 65x": "Bristol Bobber 650",
    "CFMoto 300Sx": "CFMoto 300SR",
    "Honda ADV 16x": "Honda ADV 160",
    "KTM 790 Dukx": "KTM 790 Duke",
    "Kawasaki KLX 23x": "Kawasaki KLX 230",
    "Motorstar Xplorer 250x": "Motorstar Xplorer 250R",
    "Suzuki Raider R150 Fx": "Suzuki Raider R150 Fi",
    "TVS Apache RTR 200 4x": "TVS Apache RTR 200 4V",
    "Yamaha MT-1x": "Yamaha MT-15",
    "Yamaha Serow 25x": "Yamaha Serow 250",
    "Yamaha Sniper 15x": "Yamaha Sniper 155"
}

sales["Product"] = sales["Product"].replace(
    product_name_corrections
)

print("Product names have been standardized.")

# ============================================
# CORRECT SALES DATA TYPES
# ============================================

sales["Date"] = pd.to_datetime(
    sales["Date"],
    errors="coerce"
)

sales["Unit Price"] = pd.to_numeric(
    sales["Unit Price"],
    errors="coerce"
)

sales["Quantity"] = pd.to_numeric(
    sales["Quantity"],
    errors="coerce"
)

sales["Total"] = pd.to_numeric(
    sales["Total"],
    errors="coerce"
)

print("Sales data types after conversion:")
print(sales.dtypes)

# ============================================
# REMOVE DUPLICATE SALES RECORDS
# ============================================

before = len(sales)

sales = sales.drop_duplicates()

after = len(sales)

print("Records before removing duplicates:", before)
print("Records after removing duplicates:", after)
print("Duplicate records removed:", before - after)

# ============================================
# CHECK MISSING VALUES
# ============================================

print("Missing values in Sales dataset:")
print(sales.isnull().sum)

# ============================================
# HANDLE MISSING VALUES
# ============================================

before = len(sales)

sales = sales.dropna(
    subset=[
        "Date",
        "Client Type",
        "Product",
        "Unit Price",
        "Quantity",
        "Payment"
    ]
)

after = len(sales)

print("Records before removing incomplete transactions:", before)
print("Records after removing incomplete transactions:", after)
print("Incomplete records removed:", before - after)

print("\nMissing values after handling:")
print(sales.isnull().sum())

# ============================================
# STANDARDIZE CATEGORIES
# ============================================

sales["Client Type"] = (
    sales["Client Type"]
    .str.title()
)

sales["Payment"] = (
    sales["Payment"]
    .str.title()
)

print("Client Type categories:")
print(sales["Client Type"].value_counts())

print("\nPayment categories:")
print(sales["Payment"].value_counts())

# ============================================
# MATCH SALES PRODUCTS WITH PRODUCT LIST
# ============================================

price_lookup = dict(
    zip(
        products["Product Name"],
        products["Unit Price"]
    )
)

sales["Unit Price"] = sales["Product"].map(
    price_lookup
)

print("Unmatched products:")

unmatched = sales[
    sales["Unit Price"].isnull()
]

display(
    unmatched[
        ["Product"]
    ].drop_duplicates()
)

# ============================================
# REMOVE UNMATCHED PRODUCTS
# ============================================

before = len(sales)

sales = sales.dropna(
    subset=["Unit Price"]
)

after = len(sales)

print("Unmatched product records removed:", before - after)

# ============================================
# RECALCULATE TOTAL SALES
# ============================================

sales["Unit Price"] = sales["Unit Price"].astype("int64")
sales["Quantity"] = sales["Quantity"].astype("int64")

sales["Total"] = (
    sales["Unit Price"] *
    sales["Quantity"]
).astype("int64")

print("Sales totals recalculated successfully.")

# ============================================
# STANDARDIZE DATE FORMAT
# ============================================

sales["Date"] = sales["Date"].dt.strftime(
    "%Y-%m-%d"
)

print("Date format standardized.")

display(
    sales[["Date"]].head(10)
)

# ============================================
# FINAL PRODUCT LIST STRUCTURE
# ============================================

product_clean = products[
    [
        "Product ID Number",
        "Product Name",
        "Product Type",
        "Manufacturing Year",
        "Acquisiton Year",
        "Unit Price"
    ]
].copy()

print("Clean Product List:")
display(product_clean.head(10))

# ==========================================
# ORGANIZE PRODUCT LIST REPORT
# ==========================================

product = products.rename(columns={
    "Manufacturing Year": "Date of Manufacturing",
    "Acquisiton Year": "Date of Acquisiton"
})

product = product[
    [
        "Product ID Number",
        "Product Name",
        "Product Type",
        "Unit Price",
        "Date of Manufacturing",
        "Date of Acquisiton"
    ]
]

# Make Product ID Number sequential
product["Product ID Number"] = range(1, len(product) + 1)

# Clean text
product["Product Name"] = product["Product Name"].astype(str).str.strip()
product["Product Type"] = product["Product Type"].astype(str).str.strip().str.title()

# Correct numerical formats
product["Product ID Number"] = pd.to_numeric(
    product["Product ID Number"], errors="coerce"
).astype(int)

product["Unit Price"] = pd.to_numeric(
    product["Unit Price"], errors="coerce"
).astype(int)

# Correct year formats
product["Date of Manufacturing"] = pd.to_numeric(
    product["Date of Manufacturing"], errors="coerce"
).astype("Int64")

product["Date of Acquisition"] = pd.to_numeric(
    product["Date of Acquisiton"], errors="coerce"
).astype("Int64")

# Show result
display(product.head(10))

# ==========================================
# VALIDATE PRODUCT LIST REPORT
# ==========================================

print("Columns:")
print(product.columns.tolist())

print("\nMissing values:")
print(product.isnull().sum())

print("\nDuplicate rows:")
print(product.duplicated().sum())

print("\nData types:")
print(product.dtypes)

print("\nDataset shape:")
print(product.shape)

# ============================================
# REVIEW CLEAN PRODUCT LIST
# ============================================

print("===== CLEAN PRODUCT LIST =====")

print("Rows:", product_clean.shape[0])
print("Columns:", product_clean.shape[1])

print("\nMissing values:")
print(product_clean.isnull().sum())

print("\nDuplicate rows:")
print(product_clean.duplicated().sum())

print("\nData types:")
print(product_clean.dtypes)

# ============================================
# REVIEW CLEAN SALES DATASET
# ============================================

print("===== CLEAN SALES DATASET =====")

sales_clean = sales.copy()

print("Rows:", sales_clean.shape[0])
print("Columns:", sales_clean.shape[1])

print("\nMissing values:")
print(sales_clean.isnull().sum())

print("\nDuplicate rows:")
print(sales_clean.duplicated().sum())

print("\nData types:")
print(sales_clean.dtypes)

# ============================================
# VERIFY SALES CALCULATIONS
# ============================================

sales_total_check = (
    sales_clean["Total"]
    ==
    sales_clean["Unit Price"]
    *
    sales_clean["Quantity"]
)

print(
    "Are all Sales Totals correct?",
    sales_total_check.all()
)

# ============================================
# VERIFY PRODUCT MATCHING
# ============================================

product_names = set(
    product_clean["Product Name"]
)

sales_products = set(
    sales_clean["Product"]
)

unmatched_products = (
    sales_products - product_names
)

print("Unmatched products:")
print(unmatched_products)

print(
    "\nAll sales products match Product List:",
    len(unmatched_products) == 0
)

# ============================================
# FINAL PREPROCESSING SUMMARY
# ============================================

print("==========================================")
print("       DATA PREPROCESSING COMPLETE")
print("==========================================")

print("\nPRODUCT LIST")
print("------------------------------------------")
print("Records:", len(product_clean))
print(
    "Missing values:",
    product_clean.isnull().sum().sum()
)
print(
    "Duplicate rows:",
    product_clean.duplicated().sum()
)

print("\nSALES DATASET")
print("------------------------------------------")
print("Records:", len(sales_clean))
print(
    "Missing values:",
    sales_clean.isnull().sum().sum()
)
print(
    "Duplicate rows:",
    sales_clean.duplicated().sum()
)

print(
    "\nSales totals correct:",
    (
        sales_clean["Total"]
        ==
        sales_clean["Unit Price"]
        *
        sales_clean["Quantity"]
    ).all()
)

print(
    "All products matched:",
    set(sales_clean["Product"]).issubset(
        set(product_clean["Product Name"])
    )
)

print("\n==========================================")

# ============================================
# EXPORT CLEANED DATASETS
# ============================================

product_clean.to_csv(
    "preprocessed_product_list.csv",
    index=False
)

sales_clean.to_csv(
    "preprocessed_sales.csv",
    index=False
)

print("Files successfully exported!")
print()
print("preprocessed_product_list.csv")
print("preprocessed_sales.csv")

# ==========================================
# EXPORT PREPROCESSED DATASETS AS NEW CSV
# ==========================================

# Export cleaned Product List
product.to_csv("preprocessed_product_list.csv", index=False)

# Export cleaned Sales data
sales.to_csv("preprocessed_sales.csv", index=False)

print("Both preprocessed datasets were exported successfully!")
print("1. preprocessed_product_list.csv")
print("2. preprocessed_sales.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the cleaned MotorPH dataset
product = pd.read_csv("/content/MotorPH_Products_List_2025.csv")

# Count products in each category
category_counts = (
    product["Product Type"]
    .value_counts()
    .sort_values(ascending=True)
)

# Create the horizontal bar chart
fig, ax = plt.subplots(figsize=(10, 7))

bars = ax.barh(
    category_counts.index,
    category_counts.values
)

# Add chart title and axis labels
ax.set_title(
    "MotorPH Product Distribution by Category",
    fontsize=14,
    fontweight="bold",
    pad=15
)
ax.set_xlabel("Number of Products", fontsize=11)
ax.set_ylabel("Product Category", fontsize=11)

# Display the count at the end of each bar
ax.bar_label(bars, padding=4, fontsize=9)

# Improve readability
ax.set_xlim(0, category_counts.max() + 2)
ax.grid(axis="x", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Adjust layout and save the chart
plt.tight_layout()
plt.savefig(
    "MotorPH_Product_Distribution.png",
    dpi=300,
    bbox_inches="tight"
)

# Display the chart
plt.show()
```
